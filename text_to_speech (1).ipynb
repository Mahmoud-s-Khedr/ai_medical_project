# **arabic + English**

```python
# تثبيت المكتبة
!pip install edge-tts

import edge_tts
import asyncio
from IPython.display import Audio

# قائمة الأصوات المدعومة
voices = {
    "1": {"name": "English (US)", "code": "en-US-JennyNeural", "lang": "en"},
    "2": {"name": "English (UK) ", "code": "en-GB-SoniaNeural", "lang": "en"},
    "3": {"name": "العربية (مصر) ", "code": "ar-EG-SalmaNeural", "lang": "ar"},
    "4": {"name": "العربية (السعودية)", "code": "ar-SA-ZariyahNeural", "lang": "ar"}
}

print("اختر الصوت:")
for key, v in voices.items():
    print(f"{key}. {v['name']}")

choice = input("أدخل الرقم (1-4): ")
voice_code = voices[choice]["code"]

text = input("أدخل النص (عربي أو إنجليزي حسب اختيارك): ")

async def speak():
    communicate = edge_tts.Communicate(text, voice_code)
    await communicate.save("speech.mp3")

await speak()
display(Audio("speech.mp3", autoplay=True))

# from google.colab import files
# files.download("speech.mp3")
```
